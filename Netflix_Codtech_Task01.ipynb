```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
import time
spark=SparkSession.builder.appName("Netflix Large Scale Analysis").getOrCreate()
df=spark.read.csv("netflix dataset.csv",header=True,inferSchema=True)
print("Total Rows: ",df.count())
print("Columns: ")
df.columns
df.printSchema()
df.show(5)
df=df.dropDuplicates()
df=df.na.drop()
print("\nTop Rated Movies: ")
top_movies=df.orderBy(col("rating").desc())
top_movies.select("title","rating","genre").show()
```

```python
print("\nMost popular Genres")
genre_count=df.groupBy("genre").count().orderBy("count",ascending=False)
genre_count.show(10)
```

```python
print("\nMovies Per Year")
movies_per_year = df.groupBy("year").count().orderBy("year")
movies_per_year.show()
```

```python
print("\nLongest Movies")
long_movies = df.orderBy(col("duration").desc())
long_movies.select("title","duration").show(10)
```

```python
print("\nMost Voted Movies")
vote_movies=df.orderBy(col("votes").desc())
vote_movies.select("title","votes","rating").show(10)
```

```python
print("\nPerformance Test")
start = time.time()
df.groupBy("genre").count().show()
end = time.time()
print("Execution Time:", end - start)
spark.stop()
```
